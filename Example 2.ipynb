#### PARA AÇÃO: Preencher as variáveis referentes ao ficheiro:

```python
# indicar nome do ficheiro do INE
ine="05092025_Bens_amostra fixa_INE-2025.xlsx"
ano1=2025

# pasta onde está o ficheiro
ficheiro="G:/DDEEE/05_Sub_sistemas/0511_Comercio_externo/2. Articulação INE/1. Partilha amostra fixa de empresas/3. Ficheiros INE/" + ine

# para a função ler o ficheiro, o excel tem de estar aberto com a password inserida para que esta consiga ler os dados
wb = xw.Book(ficheiro)
```

##### O código abaixo importa a folha das importações, cria o header e o fluxo associado

```python
################################
sheetI = wb.sheets['IMPORT']

# Read the entire sheet into a DataFrame
dfimp = sheetI.used_range.options(pd.DataFrame, index=False,  header=False).value

# Set row 2 as the header
dfimp.columns = dfimp.iloc[2] #pode ser necessário alterar a linha, dependendo de como vier o ficheiro do INE

# Drop rows 0, 1, 2, and 3
dfimp = dfimp.iloc[3:] #pode ser necessário alterar a linha, dependendo de como vier o ficheiro do INE


dfimp=dfimp.melt(['NPC','Dsg'])
dfimp.columns =['NPC','Dsg','Mes','contrav']
dfimp['Mes']=dfimp['Mes'].astype(str).astype(int)
dfimp['NPC']=dfimp['NPC'].astype(float).astype(int)
# dfimp['NPC'] = dfimp['NPC'].astype(str).astype(int) <- usar este se o da linha superior der erro
dfimp['Fluxo']=1

dfimp=dfimp.dropna()
dfimp
```

##### O código abaixo importa a folha das importações, cria o header e o fluxo associado

```python
################################
sheetE = wb.sheets['EXPORT']

dfexp = sheetE.used_range.options(pd.DataFrame, index=False, header=False).value

# Set row 2 as the header
dfexp.columns = dfexp.iloc[2] #pode ser necessário alterar a linha, dependendo de como vier o ficheiro do INE

# Drop rows 0, 1, 2, and 3
dfexp = dfexp.iloc[3:] #pode ser necessário alterar a linha, dependendo de como vier o ficheiro do INE

dfexp=dfexp.melt(['NPC','Dsg'])
dfexp.columns =['NPC','Dsg','Mes','contrav']
dfexp['Mes']=dfexp['Mes'].astype(str).astype(int)
dfexp['NPC']=dfexp['NPC'].astype(str).astype(int)
# dfexp['NPC'] = dfexp['NPC'].astype(str).astype(int) <- usar este se o da linha superior der erro
dfexp['Fluxo']=2

dfexp=dfexp.dropna()
dfexp
```

### Junta as duas tabelas numa e exporta-a para a pasta habitual, preparado para a integração no SAS

```python
combined_df = pd.concat([dfimp, dfexp], ignore_index=True).drop(columns='Dsg') 
combined_df['Ano'] = ano1 
combined_df.to_excel('S:/BoP/Data1/BoP/30_PRODUCAO/303_ANALISE/Comércio Externo/Comparação_INE/1. Importação dados INE/Auxiliar_importação_dados_INE_SAS_mensal.xlsx',
                     index=False,sheet_name='INE')
```
